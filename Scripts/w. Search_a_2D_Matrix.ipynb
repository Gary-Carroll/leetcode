You are given an m x n 2-D integer array matrix and an integer target.

    Each row in matrix is sorted in non-decreasing order.
    The first integer of every row is greater than the last integer of the previous row.

Return true if target exists within matrix or false otherwise.

Can you write a solution that runs in O(log(m * n)) time?

Example:

Input: matrix = [[1,2,4,8],[10,11,12,13],[14,20,30,40]]

target = 10

Output: true

```python
matrix = [[1,2,4,8],[10,11,12,13],[14,20,30,40]]
target = 10
```

```python
matrix=[[1,2,4,8],[10,11,12,13],[14,20,30,40]]
target=15
```

```python
from typing import List
class Solution:
    def searchMatrix(self, matrix: List[List[int]], target: int) -> bool:
        result = False
        target_row = []
        
        if matrix[0][0] <= target <= matrix[-1][-1]:
            for row in matrix:
                if row[0] <= target <= row[-1]:
                    target_row = row
        
        while target_row and result == False:
            half = len(target_row)//2
            if target_row[half] == target or target_row[half - 1] == target:
                result = True
            elif target > target_row[half]:
                target_row = target_row[half:]
            else:
                target_row = target_row[:half]
            if len(target_row) == 1:
                target_row = []
            
        return result
```

```python
Solution().searchMatrix(matrix, target)
```
