Implement a time-based key-value data structure that supports:

    Storing multiple values for the same key at specified time stamps
    Retrieving the key's value at a specified timestamp

Implement the TimeMap class:

    TimeMap() Initializes the object.
    void set(String key, String value, int timestamp) Stores the key key with the value value at the given time timestamp.
    String get(String key, int timestamp) Returns the most recent value of key if set was previously called on it and the most recent timestamp for that key prev_timestamp is less than or equal to the given timestamp (prev_timestamp <= timestamp). If there are no values, it returns "".

Note: For all calls to set, the timestamps are in strictly increasing order.

Example 1:

Input:
["TimeMap", "set", ["alice", "happy", 1], "get", ["alice", 1], "get", ["alice", 2], "set", ["alice", "sad", 3], "get", ["alice", 3]]

Output:
[null, null, "happy", "happy", null, "sad"]

Explanation:

TimeMap timeMap = new TimeMap();<br>
timeMap.set("alice", "happy", 1);  // store the key "alice" and value "happy" along with timestamp = 1.<br>
timeMap.get("alice", 1);           // return "happy"<br>
timeMap.get("alice", 2);           // return "happy", there is no value stored for timestamp 2, thus we return the value at timestamp 1.<br>
timeMap.set("alice", "sad", 3);    // store the key "alice" and value "sad" along with timestamp = 3.<br>
timeMap.get("alice", 3);           // return "sad"

```python
class TimeMap:
    def __init__(self):
        self.store = {}

    def set(self, key: str, value: str, timestamp: int) -> None:
        if key in self.store:
            self.store[key][0].append(value)
            self.store[key][1].append(timestamp)
        else:
            self.store[key] = ([value], [timestamp])

    def get(self, key: str, timestamp: int) -> str:
        nums = []
        if key in self.store:
            nums = self.store[key][1]
            index = 0
            while len(nums) > 1:
                half = len(nums)//2
                left, right = nums[:half], nums[half:]
                if timestamp >= right[0]:
                    nums = right
                    index += len(left)
                elif timestamp >= left[0]:
                    nums = left
                else:
                    nums = []
        return self.store[key][0][index] if nums and nums[0] <= timestamp else ''
```

```python
inputs = ["TimeMap", 
 "set", ["check", "one", 5], 
 "set", ["check", "two", 10], 
 "get", ["check", 7], 
 "get", ["nonexistent", 7]]
```
