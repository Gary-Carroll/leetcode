You are given an array of distinct integers nums, sorted in ascending order, and an integer target.<br>
Implement a function to search for target within nums. If it exists, then return its index, otherwise, return -1.<br>
Your solution must run in <i>O(logn)</i> time.

Difficulty: Easy

Input: nums = [-1,0,2,4,6,8], target = 4

Output: 3

```python
nums = [-1,0,2,4,6,8,9]
target = 4
```

```python
nums=[-1,0,3,5,9,12]
target=9
```

```python
from typing import List
class Solution:
    def search(self, nums: List[int], target: int) -> int:
        halves = []
        result = -1
        while result == -1  and nums:
            half = len(nums)//2
            if nums[half] == target: 
                result = sum(halves) + half
            elif nums[half - 1] == target:
                result = sum(halves) + half - 1
            elif target > nums[half]:
                nums = nums[half:]
                halves.append(half)
            else:
                nums = nums[:half]
            if len(nums) == 1:
                nums = []
        return result
```

```python
Solution().search(nums, target)
```
